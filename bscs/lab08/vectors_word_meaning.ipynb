```python
import pandas as pd
import numpy as np
from collections import Counter
```

```python
import json

with open('../../data/broad_context_mixed_182MB.jsonl', 'r') as f:
    data = [json.loads(line) for line in f]
```

```python
import re
import nltk
def normalize_text_advanced(text):
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[^a-z0-9\s]', '', text)  
    text = re.sub(r'\b\d+\b', '', text)
    text = re.sub(r'\b\w{1,2}\b', '', text) 
    text = re.sub(r'\b\w{20,}\b', '', text)
    stopwords = set(nltk.corpus.stopwords.words('english'))
    text = ' '.join(word for word in text.split() if word not in stopwords)
    
    text = text.strip() 
    text = re.sub(r'\s+', ' ', text) 
    return text
```

```python
normalized_data = [normalize_text_advanced(item['text']) for item in data]
display(pd.DataFrame(normalized_data, columns=['normalized_text']).head(10))
```

```python
import nltk
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

def tokenize_text_advanced(text):
    tokens = word_tokenize(text)

    tokens = [
        cleaned
        for token in tokens
        if (cleaned := re.sub(r'[^a-z0-9]', '', token.lower()))
        and cleaned not in stop_words
    ]

    return tokens

tokenized_data = [
    tokenize_text_advanced(text)
    for text in normalized_data
]

print(tokenized_data[:10])
```

```python
class CountVectorizer:
    def __init__(self):
        self.vocabulary_ = {}

    def fit(self, documents):
        unique_words = sorted(set(
            word for doc in documents for word in doc
        ))

        self.vocabulary_ = {
            word: idx for idx, word in enumerate(unique_words)
        }
        return self

    def transform(self, documents):
        count_vectors = np.zeros(
            (len(documents), len(self.vocabulary_)),
            dtype=int
        )

        for i, doc in enumerate(documents):
            word_counts = Counter(doc)

            for word, count in word_counts.items():
                if word in self.vocabulary_:
                    count_vectors[i, self.vocabulary_[word]] = count

        return count_vectors

    def fit_transform(self, documents):
        self.fit(documents)
        return self.transform(documents)

    def get_feature_names(self):
        return list(self.vocabulary_.keys())


countVectorizer = CountVectorizer()

count_vectors = countVectorizer.fit_transform(tokenized_data)

count_vectors_df = pd.DataFrame(
    count_vectors,
    columns=countVectorizer.get_feature_names()
)

print(count_vectors_df.head())
```

```python
word_vector_participants = count_vectors_df['participants'].values
word_vector_participant = count_vectors_df['participant'].values

word_vector_work = count_vectors_df['work'].values
word_vector_jobs = count_vectors_df['jobs'].values

word_vector_fictional = count_vectors_df['fictional'].values
word_vector_invented = count_vectors_df['invented'].values
word_vector_real = count_vectors_df['real'].values

word_vector_review = count_vectors_df['review'].values
word_vector_evaluation = count_vectors_df['evaluation'].values
```

```python
def cosine_similarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1)
    norm_vec2 = np.linalg.norm(vec2)
    return dot_product / (norm_vec1 * norm_vec2)
```

```python
print(f"COS(participants, participant): {cosine_similarity(word_vector_participants, word_vector_participant)}")
print(f"COS(work, jobs): {cosine_similarity(word_vector_work, word_vector_jobs)}")
print(f"COS(fictional, invented): {cosine_similarity(word_vector_fictional, word_vector_invented)}")
print(f"COS(fictional, real): {cosine_similarity(word_vector_fictional, word_vector_real)}")
print(f"COS(review, evaluation): {cosine_similarity(word_vector_review, word_vector_evaluation)}")
```

As count vectorizer does not contain negative entries in the vectors, we cannot measure fair similarity between words. Opposite words had shown as similiar. Cos(work, jobs) got to be 0 due to the fact that jobs does not appear in the context of work. 

```python
from scipy.sparse import csr_matrix, triu
from sklearn.preprocessing import normalize

ranker = Counter()

X = csr_matrix(count_vectors_df.to_numpy())

X_norm = normalize(X, norm="l2", axis=0)

similarity_matrix = X_norm.T @ X_norm

similarity_matrix = triu(similarity_matrix, k=1).tocoo()

words = count_vectors_df.columns.to_numpy()

for i, j, similarity in zip(
    similarity_matrix.row,
    similarity_matrix.col,
    similarity_matrix.data
):
    if 1.0 - 1e-12 > similarity > 0.5:
        ranker[(words[i], words[j])] = similarity

top_50 = ranker.most_common(50)
display(pd.DataFrame(top_50, columns=['Word Pair', 'Cosine Similarity']))
```
